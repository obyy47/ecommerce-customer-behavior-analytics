E-Commerce Customer Behavior Analysis

This project analyzes customer behavior patterns, spending habits, satisfaction levels, and membership performance using Python for exploratory data analysis (EDA).

Import Libraries

Importing Python libraries for data manipulation and visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Load Dataset

Loading the cleaned e-commerce customer behavior dataset into a Pandas DataFrame.

```python
df = pd.read_csv('dataset/E-commerce Customer Behavior_final.csv')
```

Preview Dataset

Displaying the first few rows of the dataset for initial inspection.

```python
df.head()
```

Dataset Information

Checking dataset structure, column names, data types, and non-null values.

```python
df.info()
```

Data Type Standardization

Converting columns into appropriate data types for analysis and visualization.

```python
df['Customer_ID'] = df['Customer_ID'].astype(str)
df['Gender'] = df['Gender'].astype(str)
df['Age'] = df['Age'].astype(int)
df['City'] = df['City'].astype(str)
df['Membership_Type'] = df['Membership_Type'].astype(str)
df['Total_Spend'] = df['Total_Spend'].astype(float)
df['Items_Purchased'] = df['Items_Purchased'].astype(int)
df['Average_Rating'] = df['Average_Rating'].astype(float)
df['Discount_Applied'] = df['Discount_Applied'].astype(bool)
df['Days_Since_Last_Purchase'] = df['Days_Since_Last_Purchase'].astype(int)
df['Satisfaction_Level'] = df['Satisfaction_Level'].astype(str)
```

Recheck Dataset Information

Verifying that all columns have been converted into the correct data types.

```python
df.info()
```

Statistical Summary

Generating descriptive statistics for numerical columns.

```python
df.describe()
```

Missing Value Check

Checking whether the dataset contains missing or null values.

```python
df.isnull().sum()
```

Distribution of Total Spend

Analyzing customer spending distribution across the dataset.

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Total_Spend'], bins=30)

plt.title('Distribution of Total Spend')
plt.xlabel('Total Spend')
plt.ylabel('Frequency')

plt.show()
```

Age Distribution

Visualizing the age distribution of customers.

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Age'], bins=20)

plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.show()
```

Membership Type vs Total Spend

Comparing customer spending behavior across different membership types.

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='Membership_Type',
    y='Total_Spend',
    data=df
)

plt.title('Membership Type vs Total Spend')

plt.show()
```

Customer Satisfaction Distribution

Analyzing the distribution of customer satisfaction levels.

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x='Satisfaction_Level',
    data=df
)

plt.title('Customer Satisfaction Level')

plt.show()
```

Discount Applied vs Total Spend

Evaluating how discounts influence customer spending behavior.

```python
plt.figure(figsize=(6,4))

sns.boxplot(
    x='Discount_Applied',
    y='Total_Spend',
    data=df
)

plt.title('Discount Applied vs Total Spend')

plt.show()
```

Average Customer Rating

Calculating the overall average customer rating score.

```python
average_star = df['Average_Rating'].mean()

print("Average Customer Rating:", round(average_star, 2))
```

Distribution of Customer Ratings

Visualizing the distribution of customer rating scores.

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['Average_Rating'],
    bins=10
)

plt.title('Distribution of Customer Ratings')
plt.xlabel('Average Rating')
plt.ylabel('Frequency')

plt.show()
```

Average Rating by Membership Type

Comparing customer satisfaction ratings across membership categories.

```python
membership_rating = df.groupby('Membership_Type')['Average_Rating'] \
                      .mean()

print(membership_rating)
```

```python
membership_rating.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Average Rating by Membership Type')
plt.xlabel('Membership Type')
plt.ylabel('Average Rating')

plt.show()
```

Rating vs Total Spend

Analyzing the relationship between customer ratings and spending behavior.

```python
sns.scatterplot(
    x='Average_Rating',
    y='Total_Spend',
    data=df
)

plt.title('Average Rating vs Total Spend')

plt.show()
```

Top Cities by Total Spend

Identifying cities that contribute the highest customer spending.

```python
top_city = df.groupby('City')['Total_Spend'] \
             .sum() \
             .sort_values(ascending=False)

print(top_city.head(10))
```

Top 10 Cities Visualization

Visualizing the highest-spending cities based on total customer spend.

```python
top_city.head(10).plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Top 10 Cities by Total Spend')
plt.xlabel('City')
plt.ylabel('Total Spend')

plt.show()
```

Gender Distribution

Analyzing customer distribution based on gender.

```python
sns.countplot(
    x='Gender',
    data=df
)

plt.title('Gender Distribution')

plt.show()
```

Membership Type Distribution

Visualizing the distribution of customers across membership categories.

```python
sns.countplot(
    x='Membership_Type',
    data=df
)

plt.title('Membership Type Distribution')

plt.show()
```

Days Since Last Purchase Distribution

Analyzing customer recency and purchase inactivity patterns.

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['Days_Since_Last_Purchase'],
    bins=20
)

plt.title('Days Since Last Purchase Distribution')
plt.xlabel('Days')
plt.ylabel('Frequency')

plt.show()
```

Average Spend by Membership Type

Comparing the average spending value across membership categories.

```python
avg_spend = df.groupby('Membership_Type')['Total_Spend'] \
              .mean()

print(avg_spend)
```

```python
avg_spend.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Average Spend by Membership Type')
plt.xlabel('Membership Type')
plt.ylabel('Average Spend')

plt.show()
```

Correlation Heatmap

Analyzing relationships between numerical customer behavior variables.

```python
correlation = df[
    [
        'Age',
        'Total_Spend',
        'Items_Purchased',
        'Average_Rating',
        'Days_Since_Last_Purchase'
    ]
].corr()

plt.figure(figsize=(7,5))

sns.heatmap(
    correlation,
    annot=True
)

plt.title('Correlation Heatmap')

plt.show()
```
